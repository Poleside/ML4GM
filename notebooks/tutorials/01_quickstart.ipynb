# ML4GM quickstart

This tutorial uses the repository's small synthetic sample dataset. It is intended to demonstrate the supported package API, not to reproduce scientific results from the full research data.

```python
from dataclasses import asdict, replace
from pathlib import Path

import pandas as pd

from ml4gm.config import RunConfig
from ml4gm.evaluation.runner import run_evaluation
```

```python
config = RunConfig.from_yaml(Path("../../configs/quickstart.yaml"))
config = replace(
    config,
    data=replace(
        config.data,
        input=Path("../../") / config.data.input,
        output_dir=Path("../../") / config.data.output_dir,
    ),
)
result = run_evaluation(config)
```

```python
fold_metrics = pd.DataFrame([asdict(fold) for fold in result.folds])
fold_metrics
```
